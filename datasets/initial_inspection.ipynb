```python
import pandas as pd

# 1. Load the datasets
df_traffic = pd.read_csv("datasets/data_mmda_traffic_spatial.csv")
df_lgu = pd.read_csv("datasets/metro_manila_local_government.csv")

# 2. Load a sample
print("Traffic Data Sample")
display(df_traffic.head(3))
print("\nLGU Data Sample")
display(df_lgu.head(3))

# 3. Report shape
print(f"\nTraffic Data Shape: {df_traffic.shape[0]} rows, {df_traffic.shape[1]} columns")
print(f"LGU Data Shape: {df_lgu.shape[0]} rows, {df_lgu.shape[1]} columns")

# 4. Display columns and types
print("\nTraffic Data Columns and Types")
df_traffic.info()
print("\nLGU Data Columns and Types")
df_lgu.info()

# 5. Report basic quality issues (Missing Values & Duplicates)
print("\nQuality Issues: Missing Values")
# Filtering to only show columns that actually have missing values
print("Traffic Missing Values:\n", df_traffic.isnull().sum()[df_traffic.isnull().sum() > 0])
print("\nLGU Missing Values:\n", df_lgu.isnull().sum()[df_lgu.isnull().sum() > 0])

print("\nQuality Issues: Duplicates")
print(f"Traffic Duplicate Rows: {df_traffic.duplicated().sum()}")
print(f"LGU Duplicate Rows: {df_lgu.duplicated().sum()}")
```
